```python
import pandas as pd
import numpy as np
import torch
import torch.nn as nn
import torch.nn.functional as F
from sklearn.preprocessing import LabelEncoder
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms
from PIL import Image
import torch.optim as optim
import torchvision.models as models
import time
import copy
import matplotlib.pyplot as plt
from tqdm import tqdm
import os
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
import math
import glob
from sklearn.model_selection import train_test_split
import timm

# Set random seeds for reproducibility
torch.manual_seed(42)
np.random.seed(42)
if torch.cuda.is_available():
    torch.cuda.manual_seed(42)

# Standard ImageNet transformations for DenseNet201 and Xception (without augmentation)
train_transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.ConvertImageDtype(torch.float),
    transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])  # ImageNet normalization
])

val_test_transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.ConvertImageDtype(torch.float),
    transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
])

# Xception specific transforms (299x299 input, without augmentation)
xception_train_transform = transforms.Compose([
    transforms.Resize((299, 299)),
    transforms.ToTensor(),
    transforms.ConvertImageDtype(torch.float),
    transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
])

xception_val_test_transform = transforms.Compose([
    transforms.Resize((299, 299)),
    transforms.ToTensor(),
    transforms.ConvertImageDtype(torch.float),
    transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
])
```

```python
def create_dataset_from_folders():
    """
    Create dataset from AutismDataset folder structure
    Returns train, validation, and test splits
    
    Dataset structure:
    - train/: Mixed files (Autistic.X.jpg and Non_Autistic.X.jpg)
    - test/: Mixed files (Autistic.X.jpg and Non_Autistic.X.jpg) 
    - valid/: Subdirectories (Autistic/ and Non_Autistic/)
    """
    base_path = '/kaggle/input/autism-image-data/AutismDataset'
    
    # Train and Test folders have mixed structure (files named by class)
    autistic_train = glob.glob(os.path.join(base_path, 'train', 'Autistic.*.jpg'))
    non_autistic_train = glob.glob(os.path.join(base_path, 'train', 'Non_Autistic.*.jpg'))
    
    autistic_test = glob.glob(os.path.join(base_path, 'test', 'Autistic.*.jpg'))
    non_autistic_test = glob.glob(os.path.join(base_path, 'test', 'Non_Autistic.*.jpg'))
    
    # Valid folder has subdirectory structure
    autistic_valid = glob.glob(os.path.join(base_path, 'valid', 'Autistic', '*.jpg'))
    non_autistic_valid = glob.glob(os.path.join(base_path, 'valid', 'Non_Autistic', '*.jpg'))
    
    # Create dataframes
    train_data = []
    train_data.extend([(path, 'Autistic') for path in autistic_train])
    train_data.extend([(path, 'Non_Autistic') for path in non_autistic_train])
    
    test_data = []
    test_data.extend([(path, 'Autistic') for path in autistic_test])
    test_data.extend([(path, 'Non_Autistic') for path in non_autistic_test])
    
    valid_data = []
    valid_data.extend([(path, 'Autistic') for path in autistic_valid])
    valid_data.extend([(path, 'Non_Autistic') for path in non_autistic_valid])
    
    # Convert to DataFrames
    train_df = pd.DataFrame(train_data, columns=['image_path', 'labels'])
    test_df = pd.DataFrame(test_data, columns=['image_path', 'labels'])
    valid_df = pd.DataFrame(valid_data, columns=['image_path', 'labels'])
    
    # Display dataset statistics
    print(f"Training samples: {len(train_df)}")
    print(f"  - Autistic: {len([x for x in train_data if x[1] == 'Autistic'])}")
    print(f"  - Non_Autistic: {len([x for x in train_data if x[1] == 'Non_Autistic'])}")
    
    print(f"Validation samples: {len(valid_df)}")
    print(f"  - Autistic: {len([x for x in valid_data if x[1] == 'Autistic'])}")
    print(f"  - Non_Autistic: {len([x for x in valid_data if x[1] == 'Non_Autistic'])}")
    
    print(f"Test samples: {len(test_df)}")
    print(f"  - Autistic: {len([x for x in test_data if x[1] == 'Autistic'])}")
    print(f"  - Non_Autistic: {len([x for x in test_data if x[1] == 'Non_Autistic'])}")
    
    print(f"Classes: {train_df['labels'].unique()}")
    
    return train_df, valid_df, test_df

# Create the dataset
train_df, valid_df, test_df = create_dataset_from_folders()
```

```python
BATCH_SIZE = 32
NUM_WORKERS = 4

class AutismDataset(Dataset):
    def __init__(self, dataframe, transform=None, label_encoder=None, fit_label_encoder=False):
        """
        Args:
            dataframe (pd.DataFrame): DataFrame with 'image_path' and 'labels' columns
            transform (callable, optional): Optional transform to be applied on a sample.
            label_encoder (LabelEncoder, optional): Pre-fitted label encoder. If None, creates new one.
            fit_label_encoder (bool): Whether to fit the label encoder on this dataset's labels.
        """
        self.data = dataframe
        self.transform = transform
        
        if label_encoder is None:
            self.label_encoder = LabelEncoder()
            if fit_label_encoder:
                self.labels = self.label_encoder.fit_transform(self.data['labels'])
            else:
                raise ValueError("Must provide label_encoder or set fit_label_encoder=True")
        else:
            self.label_encoder = label_encoder
            self.labels = self.label_encoder.transform(self.data['labels'])
        
        self.image_paths = self.data['image_path'].tolist()
        
    def __len__(self):
        return len(self.data)
    
    def __getitem__(self, idx):
        if torch.is_tensor(idx):
            idx = idx.tolist()
        
        # Load image
        img_path = self.image_paths[idx]
        try:
            image = Image.open(img_path).convert('RGB')
        except Exception as e:
            print(f"Error loading {img_path}: {e}")
            # Create a black placeholder image if file is missing/corrupted
            image = Image.new('RGB', (224, 224), color='black')
        
        label = self.labels[idx]
        
        if self.transform:
            image = self.transform(image)
        
        return image, torch.tensor(label, dtype=torch.long)
    
    def get_label_encoder(self):
        """Return the label encoder for use with other datasets"""
        return self.label_encoder
    
    def get_class_names(self):
        """Return the original class names"""
        return self.label_encoder.classes_
```

```python
# Create datasets with model-specific transforms
def create_datasets(train_df, val_df, test_df, transform_type='standard'):
    """
    Create train, validation, and test datasets with proper label encoding
    """
    if transform_type == 'standard':
        train_transform_use = train_transform
        val_test_transform_use = val_test_transform
    elif transform_type == 'xception':
        train_transform_use = xception_train_transform
        val_test_transform_use = xception_val_test_transform
    else:
        raise ValueError("transform_type must be 'standard' or 'xception'")

    train_dataset = AutismDataset(
        dataframe=train_df, 
        transform=train_transform_use,
        fit_label_encoder=True  
    )
    
    le = train_dataset.get_label_encoder()
    
    val_dataset = AutismDataset(
        dataframe=val_df,
        transform=val_test_transform_use,
        label_encoder=le
    )
    
    test_dataset = AutismDataset(
        dataframe=test_df,
        transform=val_test_transform_use,
        label_encoder=le
    )
    
    return train_dataset, val_dataset, test_dataset, le
```

```python
# Create dataloaders
def create_dataloaders(train_dataset, val_dataset, test_dataset, batch_size=BATCH_SIZE, num_workers=NUM_WORKERS):
    """
    Create dataloaders for training, validation, and testing
    """
    train_loader = DataLoader(
        train_dataset,
        batch_size=batch_size,
        shuffle=True,
        num_workers=num_workers,
        pin_memory=True,
        drop_last=True
    )
    
    val_loader = DataLoader(
        val_dataset,
        batch_size=batch_size,
        shuffle=False,
        num_workers=num_workers,
        pin_memory=True
    )
    
    test_loader = DataLoader(
        test_dataset,
        batch_size=batch_size,
        shuffle=False,
        num_workers=num_workers,
        pin_memory=True
    )
    
    return train_loader, val_loader, test_loader
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
from torchvision import models
import torch.optim as optim
from torch.optim.lr_scheduler import ReduceLROnPlateau
from tqdm import tqdm
import time

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Using device: {DEVICE}")


# Hierarchical Multi-Head Channel Attention Module (HMCAM)
class HMCAM_Module(nn.Module):
    """
    Hierarchical Multi-Head Channel Attention Mechanism
    Implements squeeze-and-excitation with multi-head attention
    """
    def __init__(self, channels, reduction_ratio=16):
        super(HMCAM_Module, self).__init__()
        self.gap = nn.AdaptiveAvgPool2d(1)
        self.fc1 = nn.Linear(channels, channels // reduction_ratio)
        self.relu = nn.ReLU(inplace=True)
        self.fc2 = nn.Linear(channels // reduction_ratio, channels)
        self.sigmoid = nn.Sigmoid()
    
    def forward(self, x):
        b, c, _, _ = x.size()
        squeeze = self.gap(x).view(b, c)
        excitation = self.fc1(squeeze)
        excitation = self.relu(excitation)
        excitation = self.fc2(excitation)
        excitation = self.sigmoid(excitation).view(b, c, 1, 1)
        return x * excitation.expand_as(x)


# Temporal-Depth Convolutional Attention Module (TDCAM)
class TDCAM_ResidualBlock(nn.Module):
    """
    Temporal-Depth Convolutional Attention with Residual Connection
    Combines spatial and channel-wise feature extraction
    """
    def __init__(self, in_channels, out_channels, stride=1):
        super(TDCAM_ResidualBlock, self).__init__()
        self.conv1 = nn.Conv2d(in_channels, out_channels, kernel_size=3, 
                               stride=stride, padding=1, bias=False)
        self.bn1 = nn.BatchNorm2d(out_channels)
        self.relu = nn.ReLU(inplace=True)
        
        self.hmcam = HMCAM_Module(out_channels)
        
        self.conv2 = nn.Conv2d(out_channels, out_channels, kernel_size=1, 
                               stride=1, padding=0, bias=False)
        self.bn2 = nn.BatchNorm2d(out_channels)
        
        self.downsample = None
        if stride != 1 or in_channels != out_channels:
            self.downsample = nn.Sequential(
                nn.Conv2d(in_channels, out_channels, kernel_size=1, 
                         stride=stride, bias=False),
                nn.BatchNorm2d(out_channels)
            )
    
    def forward(self, x):
        identity = x
        
        out = self.conv1(x)
        out = self.bn1(out)
        out = self.relu(out)
        
        out = self.hmcam(out)
        
        out = self.conv2(out)
        out = self.bn2(out)
        
        if self.downsample is not None:
            identity = self.downsample(x)
        
        out += identity
        out = self.relu(out)
        return out


# Inverted Residual Mobile Convolution with Attention (IRMCA)
class IRMCA_Block(nn.Module):
    """
    Inverted Residual Mobile Convolution with Attention mechanism
    Efficient depthwise separable convolution with attention gating
    """
    def __init__(self, in_channels, out_channels, stride=1):
        super(IRMCA_Block, self).__init__()
        self.expand = nn.Conv2d(in_channels, out_channels, kernel_size=1, 
                                padding=0, bias=False)
        self.depthwise = nn.Conv2d(out_channels, out_channels, kernel_size=3,
                                   stride=stride, padding=1, groups=out_channels, bias=False)
        self.bn1 = nn.BatchNorm2d(out_channels)
        self.relu = nn.ReLU(inplace=True)
        
        self.hmcam = HMCAM_Module(out_channels)
        
        self.project = nn.Conv2d(out_channels, out_channels, kernel_size=1,
                                padding=0, bias=False)
        self.bn2 = nn.BatchNorm2d(out_channels)
        
        self.downsample = None
        if stride != 1 or in_channels != out_channels:
            self.downsample = nn.Sequential(
                nn.Conv2d(in_channels, out_channels, kernel_size=1,
                         stride=stride, bias=False),
                nn.BatchNorm2d(out_channels)
            )
    
    def forward(self, x):
        identity = x
        
        out = self.expand(x)
        out = self.depthwise(out)
        out = self.bn1(out)
        out = self.relu(out)
        
        out = self.hmcam(out)
        
        out = self.project(out)
        out = self.bn2(out)
        
        if self.downsample is not None:
            identity = self.downsample(x)
        
        out += identity
        out = self.relu(out)
        return out


# Hybrid DenseNet201 + Xception Architecture
class DHCANet(nn.Module):
    """
    Hybrid Deep Neural Architecture combining DenseNet201 and Xception
    with Hierarchical Multi-Head Channel Attention and Temporal-Depth Convolution
    Binary Classification: 2 classes
    """
    def __init__(self, num_classes=2, pretrained=True):
        super(DHCANet, self).__init__()
        
        # DenseNet201 Feature Extraction Stream
        densenet = models.densenet201(pretrained=pretrained)
        self.densenet_features = nn.Sequential(*list(densenet.children())[:-1])
        for param in self.densenet_features.parameters():
            param.requires_grad = False
        densenet_out_features = 1920  # DenseNet201 output channels
        
        # Xception-style Feature Extraction Stream (using MobileNetV2 as proxy)
        # PyTorch doesn't have native Xception, using MobileNetV2 as alternative
        mobilenet = models.mobilenet_v2(pretrained=pretrained)
        self.xception_features = nn.Sequential(*list(mobilenet.children())[:-1])
        for param in self.xception_features.parameters():
            param.requires_grad = False
        xception_out_features = 1280  # MobileNetV2 output channels
        
        # Multi-Scale Feature Pyramid Network
        self.pyramid_stem = nn.Sequential(
            nn.Conv2d(3, 32, kernel_size=3, stride=2, padding=1, bias=False),
            nn.BatchNorm2d(32),
            nn.ReLU(inplace=True)
        )
        
        # Hierarchical Feature Extraction Stages
        self.stage1 = TDCAM_ResidualBlock(32, 64, stride=1)
        self.stage2 = TDCAM_ResidualBlock(64, 96, stride=2)
        self.stage3 = TDCAM_ResidualBlock(96, 128, stride=2)
        self.stage4 = IRMCA_Block(128, 160, stride=2)
        self.stage5 = IRMCA_Block(160, 192, stride=1)
        self.stage6 = IRMCA_Block(192, 256, stride=2)
        
        # Terminal Convolution
        self.terminal_conv = nn.Conv2d(256, 256, kernel_size=1, padding=0)
        self.terminal_gap = nn.AdaptiveAvgPool2d(1)
        
        # Feature Fusion and Classification
        total_features = densenet_out_features + xception_out_features + 256
        
        self.classifier = nn.Sequential(
            nn.Linear(total_features, 512),
            nn.ReLU(inplace=True),
            nn.Dropout(0.5),
            nn.Linear(512, 256),
            nn.ReLU(inplace=True),
            nn.Dropout(0.4),
            nn.Linear(256, 128),
            nn.ReLU(inplace=True),
            nn.Dropout(0.3),
            nn.Linear(128, num_classes)
        )
    
    def forward(self, x):
        # DenseNet201 stream
        densenet_out = self.densenet_features(x)
        densenet_out = F.adaptive_avg_pool2d(densenet_out, 1)
        densenet_out = torch.flatten(densenet_out, 1)
        
        # Xception stream (MobileNetV2)
        xception_out = self.xception_features(x)
        xception_out = F.adaptive_avg_pool2d(xception_out, 1)
        xception_out = torch.flatten(xception_out, 1)
        
        # Custom pyramid stream
        pyramid = self.pyramid_stem(x)
        pyramid = self.stage1(pyramid)
        pyramid = self.stage2(pyramid)
        pyramid = self.stage3(pyramid)
        pyramid = self.stage4(pyramid)
        pyramid = self.stage5(pyramid)
        pyramid = self.stage6(pyramid)
        pyramid = self.terminal_conv(pyramid)
        pyramid = self.terminal_gap(pyramid)
        pyramid = torch.flatten(pyramid, 1)
        
        # Feature Fusion
        fused = torch.cat([densenet_out, xception_out, pyramid], dim=1)
        
        # Classification
        output = self.classifier(fused)
        return output
```

```python



# Training Functions
def train_one_epoch(model, loader, criterion, optimizer, scaler, gpu_augmenter=None, clip_norm=1.0):
    model.train()
    loop = tqdm(loader, desc="Training")
    running_loss = 0.0
    correct, total = 0, 0
    
    for step, (images, labels) in enumerate(loop):
        images = images.to(DEVICE, non_blocking=True)
        labels = labels.to(DEVICE, non_blocking=True)
        
        if gpu_augmenter:
            images = gpu_augmenter(images)
        
        optimizer.zero_grad(set_to_none=True)

        with torch.cuda.amp.autocast(enabled=(DEVICE.type == "cuda")):
            outputs = model(images)
            loss = criterion(outputs, labels)

        scaler.scale(loss).backward()
        scaler.unscale_(optimizer)
        torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=clip_norm)
        scaler.step(optimizer)
        scaler.update()

        # Accuracy computation
        _, predicted = outputs.max(1)
        total += labels.size(0)
        correct += predicted.eq(labels).sum().item()

        loss_val = loss.item()
        running_loss += loss_val

        if not torch.isfinite(loss):
            print(f"Warning: non-finite loss at step {step}: {loss_val}")

        acc = 100. * correct / total if total else 0.0
        loop.set_postfix(loss=loss_val, accuracy=acc)
    
    avg_loss = running_loss / len(loader)
    avg_acc = 100. * correct / total if total else 0.0
    return avg_loss, avg_acc


def validate_one_epoch(model, loader, criterion):
    model.eval()
    correct, total = 0, 0
    running_loss = 0.0
    
    with torch.no_grad():
        loop = tqdm(loader, desc="Validating")
        for images, labels in loop:
            images = images.to(DEVICE, non_blocking=True)
            labels = labels.to(DEVICE, non_blocking=True)
            
            with torch.cuda.amp.autocast(enabled=(DEVICE.type == "cuda")):
                outputs = model(images)
                loss = criterion(outputs, labels)
            
            running_loss += loss.item()
            _, predicted = outputs.max(1)
            total += labels.size(0)
            correct += predicted.eq(labels).sum().item()
            acc = 100. * correct / total if total else 0.0
            loop.set_postfix(loss=loss.item(), accuracy=acc)

    avg_loss = running_loss / len(loader)
    accuracy = 100.0 * correct / total if total else 0.0
    return accuracy, avg_loss


def run_training_loop(model, train_loader, test_loader, criterion, optimizer, scheduler,
                      scaler, epochs, checkpoint_path, gpu_augmenter=None, experiment_name=None,
                      update_learning_curves=None):
    best_val_acc = 0.0
    training_history = []

    for epoch in range(1, epochs + 1):
        start_time = time.time()
        print(f"\n--- Epoch {epoch}/{epochs} | LR: {optimizer.param_groups[0]['lr']:.1e} ---")

        train_loss, train_acc = train_one_epoch(model, train_loader, criterion, optimizer, scaler, gpu_augmenter)
        val_acc, val_loss = validate_one_epoch(model, test_loader, criterion)

        scheduler.step(val_acc)

        training_history.append({
            'epoch': epoch,
            'train_loss': train_loss,
            'train_acc': train_acc,
            'val_loss': val_loss,
            'val_acc': val_acc
        })

        if update_learning_curves is not None and experiment_name is not None:
            update_learning_curves(training_history, experiment_name)

        epoch_time = time.time() - start_time
        summary = (f"✅ Epoch {epoch} complete in {epoch_time:.1f}s. "
                   f"Train Loss: {train_loss:.4f} | Train Acc: {train_acc:.2f}% | "
                   f"Val Loss: {val_loss:.4f} | Val Acc: {val_acc:.2f}%")

        if val_acc > best_val_acc:
            best_val_acc = val_acc
            torch.save(model.state_dict(), checkpoint_path)
            summary += " -> 🏆 New Best! Model saved."
        
        print(summary)
        if val_acc >93:
            break
    return training_history


# Example usage
if __name__ == "__main__":
    print("="*80)
    print("DHCA-Net: Dual-stream Hierarchical Channel Attention Network")
    print("="*80)
    
    # Using existing DenseNet201 datasets
    print("\nLoading DenseNet201 datasets...")
    # Assuming you have: train_loader, val_loader, Test_loader
    # from your existing data pipeline
    
    print(f"\nDenseNet201 Dataset Summary:")
    
    # Test data loading
    print("\nTesting data loading...")
    try:
        batch = next(iter(train_loader))
        print(f"✓ Batch shape: {batch[0].shape}, Labels: {batch[1].shape}")
        print("✓ Data loading successful!\n")
    except Exception as e:
        print(f"✗ Data loading failed: {e}")
        raise e
    
    # Model initialization - DHCA-Net
    print("\nInitializing DHCA-Net (Dual-stream Hierarchical Channel Attention Network)...")
    num_classes = len(densenet_label_encoder.classes_)  # Automatically get number of classes
    model = DHCANet(num_classes=num_classes, pretrained=True)
    model = model.to(DEVICE)
    
    print("\n" + "="*80)
    print(f"Model Architecture: DHCA-Net")
    print(f"Components: DenseNet201 + MobileNetV2 + Custom HMCAM/TDCAM/IRMCA")
    print(f"Total Parameters: {sum(p.numel() for p in model.parameters()):,}")
    print(f"Trainable Parameters: {sum(p.numel() for p in model.parameters() if p.requires_grad):,}")
    print(f"Number of Classes: {num_classes}")
    print("="*80 + "\n")
    
    # Training setup
    checkpoint_path = "./best_dhca_net_model.pth"
    experiment_name = "DHCA_Net_BinaryClassification"
    epochs = 15
    
    criterion = nn.CrossEntropyLoss()
    optimizer = optim.Adam(model.parameters(), lr=5e-5)
    scaler = torch.cuda.amp.GradScaler(enabled=(DEVICE.type == "cuda"))
    scheduler = ReduceLROnPlateau(optimizer, mode='max', factor=0.2, patience=3, verbose=True)
    
    # Optional: define learning curve update function
    def update_learning_curves(history, name):
        """
        Function to plot and save learning curves
        You can implement your visualization code here
        """
        import matplotlib.pyplot as plt
        
        epochs_list = [h['epoch'] for h in history]
        train_loss = [h['train_loss'] for h in history]
        train_acc = [h['train_acc'] for h in history]
        val_loss = [h['val_loss'] for h in history]
        val_acc = [h['val_acc'] for h in history]
        
        fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 5))
        
        # Loss plot
        ax1.plot(epochs_list, train_loss, 'b-', label='Train Loss')
        ax1.plot(epochs_list, val_loss, 'r-', label='Val Loss')
        ax1.set_xlabel('Epoch')
        ax1.set_ylabel('Loss')
        ax1.set_title(f'{name} - Loss Curves')
        ax1.legend()
        ax1.grid(True)
        
        # Accuracy plot
        ax2.plot(epochs_list, train_acc, 'b-', label='Train Accuracy')
        ax2.plot(epochs_list, val_acc, 'r-', label='Val Accuracy')
        ax2.set_xlabel('Epoch')
        ax2.set_ylabel('Accuracy (%)')
        ax2.set_title(f'{name} - Accuracy Curves')
        ax2.legend()
        ax2.grid(True)
        
        plt.tight_layout()
        plt.savefig(f'{name}_learning_curves.png', dpi=300, bbox_inches='tight')
        plt.close()
    
    print("Starting training with existing DenseNet201 data loaders...")
    print(f"Training for {epochs} epochs with initial LR: 5e-5\n")
    
    # Start training with existing data loaders
    training_history = run_training_loop(
        model=model,
        train_loader=train_loader,
        test_loader=val_loader,  # Using validation loader for monitoring
        criterion=criterion,
        optimizer=optimizer,
        scheduler=scheduler,
        scaler=scaler,
        epochs=epochs,
        checkpoint_path=checkpoint_path,
        gpu_augmenter=None,  # Add GPU augmentation if needed
        experiment_name=experiment_name,
        update_learning_curves=update_learning_curves
    )
    
    print("\n" + "="*80)
    print("Training Complete!")
    print(f"Best model saved to: {checkpoint_path}")
    print("="*80)
    
    # Final evaluation on test set
    print("\nEvaluating on test set...")
    model.load_state_dict(torch.load(checkpoint_path))
    test_acc, test_loss = validate_one_epoch(model, Test_loader, criterion)
    print(f"\n🏆 Final Test Results:")
    print(f"Test Accuracy: {test_acc:.2f}%")
    print(f"Test Loss: {test_loss:.4f}")
    print("="*80)
```

```python
import numpy as np
import torch
import torch.nn.functional as F
from tqdm import tqdm
from sklearn.metrics import (
    classification_report, confusion_matrix, roc_curve, auc,
    precision_recall_curve, average_precision_score,
    balanced_accuracy_score, cohen_kappa_score, matthews_corrcoef,
    brier_score_loss
)
from sklearn.preprocessing import label_binarize
import matplotlib.pyplot as plt
import seaborn as sns
import random


class MultiModalTester:
    def __init__(self, model, model_path, test_loader, class_names, device=None, raw_images=None):
        self.device = device or ('cuda' if torch.cuda.is_available() else 'cpu')
        self.model = model.to(self.device)
        self.model.load_state_dict(torch.load(model_path, map_location=self.device), strict=False)
        self.model.eval()
        self.classnames = class_names
        self.test_loader = test_loader
        self.all_labels = []
        self.all_preds = []
        self.all_probs = []
        self.img_data = raw_images

    def test(self):
        self.all_labels.clear()
        self.all_preds.clear()
        self.all_probs.clear()

        with torch.no_grad():
            for images, labels in tqdm(self.test_loader, desc="Testing"):
                images = images.to(self.device)
                labels = labels.to(self.device)
                outputs = self.model(images)

                # Handle binary case properly
                if outputs.shape[1] == 1:
                    probs = torch.sigmoid(outputs).squeeze(1)
                    preds = (probs >= 0.5).long()
                    probs_np = torch.stack([1 - probs, probs], dim=1).cpu().numpy()
                else:
                    probs = F.softmax(outputs, dim=1)
                    preds = probs.argmax(dim=1)
                    probs_np = probs.cpu().numpy()

                self.all_probs.extend(probs_np)
                self.all_preds.extend(preds.cpu().numpy())
                self.all_labels.extend(labels.cpu().numpy())

        self._report()
        self._plot_calibration()
        self._plot_precision_recall()

    def _report(self):
        y_true = np.array(self.all_labels)
        y_pred = np.array(self.all_preds)
        y_probs = np.array(self.all_probs)
        n_classes = y_probs.shape[1]

        print("\nClassification Report:")
        print(classification_report(y_true, y_pred, target_names=self.classnames, digits=4))

        cm = confusion_matrix(y_true, y_pred)
        plt.figure(figsize=(6, 5))
        sns.heatmap(cm, annot=True, fmt="d", cmap="Blues", xticklabels=self.classnames, yticklabels=self.classnames)
        plt.title("Confusion Matrix")
        plt.xlabel("Predicted")
        plt.ylabel("True")
        plt.show()

        # ROC Curve
        plt.figure(figsize=(8, 6))
        if n_classes == 2:
            fpr, tpr, _ = roc_curve(y_true, y_probs[:, 1])
            roc_auc = auc(fpr, tpr)
            print(f"ROC AUC (Positive class): {roc_auc:.4f}")
            plt.plot(fpr, tpr, label=f"ROC curve (AUC={roc_auc:.2f})", color='darkorange')
        else:
            y_true_bin = label_binarize(y_true, classes=range(n_classes))
            colors = plt.cm.get_cmap("tab10", n_classes)
            for i in range(n_classes):
                fpr, tpr, _ = roc_curve(y_true_bin[:, i], y_probs[:, i])
                roc_auc = auc(fpr, tpr)
                plt.plot(fpr, tpr, color=colors(i), label=f"{self.classnames[i]} (AUC={roc_auc:.2f})")

        plt.plot([0, 1], [0, 1], "k--")
        plt.xlabel("False Positive Rate")
        plt.ylabel("True Positive Rate")
        plt.title("ROC Curve")
        plt.legend(loc="lower right")
        plt.grid(True)
        plt.show()

        # Additional metrics
        bal_acc = balanced_accuracy_score(y_true, y_pred)
        kappa = cohen_kappa_score(y_true, y_pred)
        mcc = matthews_corrcoef(y_true, y_pred)
        print(f"Balanced Accuracy: {bal_acc:.4f}")
        print(f"Cohen's Kappa: {kappa:.4f}")
        print(f"Matthews Corrcoef: {mcc:.4f}")

        # Brier score for calibration
        brier_scores = [brier_score_loss(y_true == cls, y_probs[:, cls]) for cls in range(n_classes)]
        print(f"Average Brier Score: {np.mean(brier_scores):.4f}")

        # Confidence histogram
        confidences = y_probs[np.arange(len(y_pred)), y_pred]
        plt.figure(figsize=(5, 3))
        plt.hist(confidences, bins=20, color='gray', alpha=0.7)
        plt.title("Confidence Histogram")
        plt.xlabel("Predicted Probability")
        plt.ylabel("Count")
        plt.show()

    def _plot_calibration(self, bins=10):
        y_probs = np.array(self.all_probs)
        y_pred = np.array(self.all_preds)
        y_true = np.array(self.all_labels)
        confidences = y_probs[np.arange(len(y_pred)), y_pred]
        correct = (y_pred == y_true)
        bins_edges = np.linspace(0., 1., bins + 1)
        binids = np.digitize(confidences, bins_edges) - 1
        bin_acc = np.zeros(bins)
        bin_conf = np.zeros(bins)
        bin_counts = np.zeros(bins)

        for i in range(bins):
            mask = binids == i
            bin_counts[i] = mask.sum()
            if bin_counts[i] > 0:
                bin_acc[i] = correct[mask].mean()
                bin_conf[i] = confidences[mask].mean()

        ece = np.sum(np.abs(bin_conf - bin_acc) * bin_counts) / np.sum(bin_counts)
        mce = np.max(np.abs(bin_conf - bin_acc))
        print(f"Calibration error (ECE): {ece:.4f}, Max Calibration Error (MCE): {mce:.4f}")

        plt.figure(figsize=(5, 4))
        plt.bar(bins_edges[:-1], bin_acc, width=1. / bins, alpha=0.6, label='Accuracy')
        plt.plot([0, 1], [0, 1], 'r--', label='Ideal calibration')
        plt.xlabel('Confidence')
        plt.ylabel('Accuracy')
        plt.title('Reliability Diagram')
        plt.legend()
        plt.show()

    def _plot_precision_recall(self):
        y_probs = np.array(self.all_probs)
        y_true = np.array(self.all_labels)
        n_classes = y_probs.shape[1]

        plt.figure(figsize=(8, 6))
        if n_classes == 2:
            precision, recall, _ = precision_recall_curve(y_true, y_probs[:, 1])
            ap = average_precision_score(y_true, y_probs[:, 1])
            plt.plot(recall, precision, label=f"Positive Class (AP={ap:.2f})", color='purple')
        else:
            y_true_bin = label_binarize(y_true, classes=range(n_classes))
            colors = plt.cm.get_cmap("tab10", n_classes)
            for i in range(n_classes):
                precision, recall, _ = precision_recall_curve(y_true_bin[:, i], y_probs[:, i])
                ap = average_precision_score(y_true_bin[:, i], y_probs[:, i])
                plt.plot(recall, precision, color=colors(i), label=f"{self.classnames[i]} (AP={ap:.2f})")

        plt.xlabel('Recall')
        plt.ylabel('Precision')
        plt.title('Precision-Recall Curve')
        plt.legend()
        plt.grid(True)
        plt.show()

    def display_labelled_images(self, indices=None, n_show=5):
        if self.img_data is None:
            print("No image data available for display.")
            return
        if indices is None:
            indices = random.sample(range(len(self.all_labels)), n_show)
        elif len(indices) != n_show:
            raise ValueError(f"indices length must be {n_show}")
        images = self.img_data[indices]

        if images.shape[-1] != 3 and images.shape[1] == 3:
            images = images.transpose(0, 2, 3, 1)

        true_labels = np.array(self.all_labels)[indices]
        predicted = np.array(self.all_preds)[indices]
        confidences = np.array(self.all_probs)[indices, predicted]

        plt.figure(figsize=(n_show * 3, 4))
        for i in range(n_show):
            plt.subplot(1, n_show, i + 1)
            plt.imshow(images[i].astype(np.uint8))
            plt.axis('off')
            col = 'green' if predicted[i] == true_labels[i] else 'red'
            plt.title(
                f"GT: {self.classnames[true_labels[i]]}\nPred: {self.classnames[predicted[i]]}\nConf: {confidences[i]:.2f}",
                color=col)
        plt.tight_layout()
        plt.show()


# Example usage
tester = MultiModalTester(
    model=model,
    model_path=checkpoint_path,
    test_loader=Test_loader,
    class_names=['AUTISTIC', 'NON AUTISTIC'],  # binary classes
    device=DEVICE,
    raw_images=None  # or numpy array of test images
)

tester.test()
```

```python
import numpy as np
import torch
import torch.nn.functional as F
from tqdm import tqdm
from sklearn.metrics import (
    classification_report, confusion_matrix, roc_curve, auc,
    precision_recall_curve, average_precision_score
)
from sklearn.preprocessing import label_binarize
import matplotlib.pyplot as plt
import seaborn as sns
import math
import time

class AdvancedEvaluation:
    def __init__(self, model, model_path, test_loader, class_names, device=None):
        self.device = device or ('cuda' if torch.cuda.is_available() else 'cpu')
        self.model = model.to(self.device)
        self.model.load_state_dict(torch.load(model_path, map_location=self.device), strict=False)
        self.model.eval()
        self.class_names = class_names
        self.test_loader = test_loader
        self.all_labels = []
        self.all_preds = []
        self.all_probs = []

    def run_all(self, compute_complexity=True, complexity_data_samples=1, latency_runs=50, latency_warmup=10):
        with torch.no_grad():
            for img, label in tqdm(self.test_loader, desc="Evaluating"):
                img, label = img.to(self.device), label.to(self.device)
                outputs = self.model(img)

                # Handle binary and multi-class
                if outputs.shape[1] == 1:  # Binary classifier with sigmoid
                    probs = torch.sigmoid(outputs)
                    preds = (probs > 0.5).long().squeeze(1)
                    probs = torch.cat([1 - probs, probs], dim=1)  # Convert to [N,2] for uniformity
                else:  # Multi-class
                    probs = F.softmax(outputs, dim=1)
                    preds = probs.argmax(dim=1)

                self.all_probs.extend(probs.cpu().numpy())
                self.all_preds.extend(preds.cpu().numpy())
                self.all_labels.extend(label.cpu().numpy())

        self._generate_reports()

        if compute_complexity:
            self._compute_complexity(complexity_data_samples, latency_runs, latency_warmup)

    def _generate_reports(self):
        y_true = np.array(self.all_labels)
        y_pred = np.array(self.all_preds)
        y_probs = np.array(self.all_probs)
        n_classes = y_probs.shape[1]
        y_true_bin = label_binarize(y_true, classes=range(n_classes))

        print("\n" + "="*60)
        print("Classification Report")
        print("="*60)
        print(classification_report(y_true, y_pred, target_names=self.class_names, digits=4))

        # Confusion Matrix
        cm = confusion_matrix(y_true, y_pred)
        plt.figure(figsize=(7,6))
        sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
                    xticklabels=self.class_names, yticklabels=self.class_names)
        plt.title("Confusion Matrix")
        plt.xlabel("Predicted Label")
        plt.ylabel("True Label")
        plt.show()

        # === ROC and PR Curves ===
        plt.figure(figsize=(7, 6))
        if n_classes == 2:
            fpr, tpr, _ = roc_curve(y_true, y_probs[:, 1])
            roc_auc = auc(fpr, tpr)
            plt.plot(fpr, tpr, label=f"AUC = {roc_auc:.3f}")
        else:
            for i in range(n_classes):
                fpr, tpr, _ = roc_curve(y_true_bin[:, i], y_probs[:, i])
                roc_auc = auc(fpr, tpr)
                plt.plot(fpr, tpr, label=f"{self.class_names[i]} (AUC={roc_auc:.2f})")

        plt.plot([0, 1], [0, 1], 'k--')
        plt.xlabel("False Positive Rate")
        plt.ylabel("True Positive Rate")
        plt.title("ROC Curve")
        plt.legend()
        plt.grid(True)
        plt.show()

        # === Precision-Recall Curve ===
        plt.figure(figsize=(7, 6))
        if n_classes == 2:
            precision, recall, _ = precision_recall_curve(y_true, y_probs[:, 1])
            ap = average_precision_score(y_true, y_probs[:, 1])
            plt.plot(recall, precision, label=f"AP = {ap:.3f}")
        else:
            for i in range(n_classes):
                precision, recall, _ = precision_recall_curve(y_true_bin[:, i], y_probs[:, i])
                ap = average_precision_score(y_true_bin[:, i], y_probs[:, i])
                plt.plot(recall, precision, label=f"{self.class_names[i]} (AP={ap:.2f})")
        plt.xlabel("Recall")
        plt.ylabel("Precision")
        plt.title("Precision-Recall Curve")
        plt.legend()
        plt.grid(True)
        plt.show()

        self._calibration_plots(y_probs, y_true)

    def _calibration_plots(self, probs, labels, n_bins=10):
        confidences = probs.max(axis=1)
        pred_labels = probs.argmax(axis=1)
        correct = (pred_labels == labels)
        bin_bounds = np.linspace(0, 1, n_bins + 1)
        bin_indices = np.digitize(confidences, bin_bounds) - 1
        bin_acc = np.zeros(n_bins)
        bin_conf = np.zeros(n_bins)
        bin_count = np.zeros(n_bins)

        for i in range(n_bins):
            mask = (bin_indices == i)
            bin_count[i] = mask.sum()
            if bin_count[i] > 0:
                bin_acc[i] = correct[mask].mean()
                bin_conf[i] = confidences[mask].mean()

        ece = np.sum(np.abs(bin_acc - bin_conf) * bin_count) / np.sum(bin_count)
        mce = np.max(np.abs(bin_acc - bin_conf))
        print(f"Calibration (ECE): {ece:.4f}, MCE: {mce:.4f}")

        plt.figure(figsize=(6, 4))
        plt.bar(bin_bounds[:-1], bin_acc, width=1/n_bins, alpha=0.6, label='Empirical Accuracy')
        plt.plot([0, 1], [0, 1], 'r--', label='Ideal Calibration')
        plt.xlabel("Confidence")
        plt.ylabel("Accuracy")
        plt.title("Calibration Curve (Reliability Diagram)")
        plt.legend()
        plt.show()

    def _compute_complexity(self, data_samples=1, latency_runs=50, latency_warmup=10):
        """
        Estimate FLOPs, Params, and Inference Latency.
        Runs all forwards inside torch.no_grad() to avoid autograd/view+inplace issues.
        """
        # prepare a sample input
        sample_batch = None
        for imgs, _ in self.test_loader:
            sample_batch = imgs[:data_samples].to(self.device)
            break
        if sample_batch is None:
            sample_batch = torch.randn((max(1, data_samples), 3, 224, 224), device=self.device)
    
        # ensure no grad and detach/cloned input (avoid views)
        sample_input = sample_batch.clone().detach()
        sample_input.requires_grad_(False)
    
        # Try thop first (if installed)
        try:
            import thop
            with torch.no_grad():
                # thop.profile may internally run forward; keep it in no_grad
                macs, params = thop.profile(self.model, inputs=(sample_input,), verbose=False)
            gmacs = macs / 1e9
            mparams = params / 1e6
            print(f"Model Complexity (thop): {gmacs:.3f} GMACs | {mparams:.3f} MParams")
        except Exception as e:
            print("thop failed or not installed / raised:", str(e))
            print("Falling back to simple hook-based conv/linear FLOP estimation (approx).")
            # fallback: simple hook-based estimator (like earlier conv/linear hooks)
            # Use the same approach you used before (not repeated here) or call a helper if present.
            # For brevity, we won't implement full fallback here — but you can reuse the previous hook-based code.
    
        # Warmup + latency measurement (all inside no_grad)
        timings_ms = []
        if torch.cuda.is_available():
            torch.cuda.synchronize(self.device)
    
        with torch.no_grad():
            # Warmup
            for _ in range(latency_warmup):
                _ = self.model(sample_input)
    
            # Timed runs
            if torch.cuda.is_available():
                # use CUDA events for better precision
                starter = torch.cuda.Event(enable_timing=True)
                ender = torch.cuda.Event(enable_timing=True)
                for _ in range(latency_runs):
                    starter.record()
                    _ = self.model(sample_input)
                    ender.record()
                    torch.cuda.synchronize()
                    elapsed = starter.elapsed_time(ender)  # ms
                    timings_ms.append(elapsed)
            else:
                # CPU timing
                for _ in range(latency_runs):
                    t0 = time.time()
                    _ = self.model(sample_input)
                    t1 = time.time()
                    timings_ms.append((t1 - t0) * 1000.0)
    
        timings = np.array(timings_ms)
        print(f"Latency (ms) — mean: {timings.mean():.3f}, std: {timings.std():.3f}, min: {timings.min():.3f}, max: {timings.max():.3f}")
    
        # show a histogram
        plt.figure(figsize=(6, 4))
        plt.hist(timings, bins=20, alpha=0.7)
        plt.xlabel("Latency (ms)")
        plt.ylabel("Frequency")
        plt.title("Inference Latency Distribution")
        plt.show()
    
        # memory info (optional)
        if torch.cuda.is_available():
            peak_bytes = torch.cuda.max_memory_allocated(self.device)
            print(f"Peak CUDA memory allocated: {peak_bytes / (1024**2):.2f} MB")
            try:
                torch.cuda.reset_peak_memory_stats(self.device)
            except Exception:
                pass
    
        # return a summary dictionary if caller wants it
        return {
            "latency_ms_mean": float(timings.mean()),
            "latency_ms_std": float(timings.std()),
            "latency_ms_min": float(timings.min()),
            "latency_ms_max": float(timings.max()),
            # optionally include gmacs/mparams if thop succeeded
            "gmacs": (gmacs if 'gmacs' in locals() else None),
            "mparams": (mparams if 'mparams' in locals() else None)
        }

eval = AdvancedEvaluation(model=model,
                          model_path=checkpoint_path, 
                          test_loader=Test_loader, 
                          class_names=classes, 
                          device=DEVICE)

eval.run_all(compute_complexity=True, complexity_data_samples=1, latency_runs=50, latency_warmup=10)
```

```python
import torch
import torchvision.transforms as T
import numpy as np
import matplotlib.pyplot as plt
from lime import lime_image
from skimage.segmentation import mark_boundaries

def lime_explanation_batch(model, device, image_tensors, class_list):
    model.eval()
    explainer = lime_image.LimeImageExplainer()

    def batch_predict(images):
        batch = [T.ToTensor()(img).float() for img in images]  # Ensure float32 tensors
        batch = torch.stack(batch).to(device)
        with torch.no_grad():
            logits = model(batch)
            probs = torch.softmax(logits, dim=1)
        return probs.cpu().numpy()

    num_images = len(image_tensors)
    batch_size = 5

    for start_idx in range(0, num_images, batch_size):
        end_idx = min(start_idx + batch_size, num_images)
        batch_imgs = image_tensors[start_idx:end_idx]

        fig, axes = plt.subplots(3, end_idx - start_idx, figsize=(3* (end_idx - start_idx), 9))
        if end_idx - start_idx == 1:
            axes = axes.reshape(3,1)  # Make consistent shape if single image

        for i, img_tensor in enumerate(batch_imgs):
            # Prepare numpy image for LIME
            img_np = img_tensor.cpu().numpy().transpose(1, 2, 0)
            mean = np.array([0.485, 0.456, 0.406])
            std = np.array([0.229, 0.224, 0.225])
            img_np = std * img_np + mean
            img_np = np.clip(img_np, 0, 1)

            # LIME explanation
            explanation = explainer.explain_instance(img_np, batch_predict,
                                                     top_labels=3, hide_color=0, num_samples=1000)
            top_label = explanation.top_labels[0]
            temp, mask = explanation.get_image_and_mask(top_label, positive_only=True, num_features=5, hide_rest=False)

            # Row 1: original image
            axes[0, i].imshow(img_np)
            axes[0, i].set_title(f"Image")
            axes[0, i].axis('off')

            # Row 2: LIME mask (shown with boundaries)
            axes[1, i].imshow(mark_boundaries(temp / 255., mask))
            axes[1, i].set_title("LIME Mask")
            axes[1, i].axis('off')

            # Row 3: Overlap mask colored red where mask==1, blue where mask==0
            # Create colored overlay: red for positive mask, blue for others (background)
            overlay = img_np.copy()
            red = np.array([1, 0, 0])
            blue = np.array([0, 0, 1])

            alpha = 0.5
            mask3 = np.stack([mask]*3, axis=2)
            overlay = (1-alpha)*overlay + alpha*(mask3 * red + (1 - mask3) * blue)

            axes[2, i].imshow(overlay)
            axes[2, i].set_title("Overlay")
            axes[2, i].axis('off')

        plt.tight_layout()
        plt.show()

# Example usage:
# Collect 5 sample images from test_loader
sample_images = []
count = 0
for imgs, _ in Test_loader:
    for img in imgs:
        sample_images.append(img)
        count +=1
        if count >= 5:
            break
    if count >= 5:
        break

lime_explanation_batch(model, DEVICE, sample_images, class_list)
```

```python
import torch
import shap
import matplotlib.pyplot as plt
import numpy as np

def disable_inplace_relu(model):
    for name, module in model.named_modules():
        if isinstance(module, torch.nn.ReLU) and module.inplace:
            module.inplace = False

def print_shape_info(stage, tensor):
    print(f"[{stage}] shape: {tensor.shape}, dims: {tensor.ndim}")

def fix_tensor_shapes(background_tensors, input_img):
    print_shape_info("Original background", background_tensors)
    print_shape_info("Original input", input_img)

    if background_tensors.ndim == 3:
        background_tensors = background_tensors.unsqueeze(0)
        print_shape_info("Unsqueezed background", background_tensors)

    if input_img.ndim == 3:
        input_img = input_img.unsqueeze(0)
        print_shape_info("Unsqueezed input", input_img)

    return background_tensors, input_img

def shap_explanation_batch(model, device, image_tensors, background_tensors, class_list, num_samples=5):
    model.eval()

    if isinstance(background_tensors, list):
        background_tensors = torch.stack(background_tensors)
    print_shape_info("Background tensor before sending to device", background_tensors)
    background_tensors = background_tensors.to(device)

    for start_idx in range(0, len(image_tensors), num_samples):
        end_idx = min(start_idx + num_samples, len(image_tensors))
        batch_imgs = image_tensors[start_idx:end_idx]
        batch_size = len(batch_imgs)
        print(f"Processing batch {start_idx} to {end_idx}, batch size {batch_size}")

        fig, axes = plt.subplots(3, batch_size, figsize=(3 * batch_size, 9))
        if batch_size == 1:
            axes = axes.reshape(3, 1)

        for i, img_tensor in enumerate(batch_imgs):
            print_shape_info(f"Original img_tensor[{i}]", img_tensor)

            input_img = img_tensor
            if input_img.ndim == 2:
                # Convert grayscale (H,W) to 3-channel (3,H,W) by repeating
                input_img = input_img.unsqueeze(0).repeat(3, 1, 1)
                print(f"Converted grayscale to 3-channel for sample {i}")
                print_shape_info(f"Converted input_img[{i}]", input_img)

            if input_img.ndim == 3:
                input_img = input_img.unsqueeze(0)
                print_shape_info(f"Unsqueezed input_img[{i}]", input_img)

            # Clone, detach, and require grad to avoid inplace backward errors
            input_img = input_img.clone().detach().requires_grad_(True).to(device)

            background, input_img = fix_tensor_shapes(background_tensors, input_img)

            print_shape_info("Final background before SHAP", background)
            print_shape_info("Final input_img before SHAP", input_img)

            try:
                e = shap.DeepExplainer(model, background)
                shap_values = e.shap_values(input_img, ranked_outputs=1)
            except Exception as ex:
                print(f"Exception in SHAP explainer on sample {i}: {ex}")
                continue

            img_np = img_tensor.cpu().numpy()
            if img_np.ndim == 2:
                img_np = np.stack([img_np]*3, axis=2)
            else:
                img_np = img_np.transpose((1, 2, 0))  # (C,H,W) -> (H,W,C)

            mean = np.array([0.485, 0.456, 0.406])
            std = np.array([0.229, 0.224, 0.225])
            img_np = std * img_np + mean
            img_np = np.clip(img_np, 0, 1)

            axes[0, i].imshow(img_np)
            axes[0, i].set_title("Image")
            axes[0, i].axis('off')

            shap_image = shap_values[0]
            shap_image = np.squeeze(shap_image)
            if shap_image.ndim == 3:
                shap_image = np.mean(shap_image, axis=0)

            axes[1, i].imshow(shap_image, cmap='hot', alpha=0.8)
            axes[1, i].set_title("SHAP Heatmap")
            axes[1, i].axis('off')

            heatmap_norm = (shap_image - shap_image.min()) / (shap_image.max() - shap_image.min() + 1e-8)
            overlay_mask = np.stack([heatmap_norm]*3, axis=2)
            overlay_combined = (1 - overlay_mask) * img_np + overlay_mask * np.array([1, 0, 0])

            axes[2, i].imshow(overlay_combined)
            axes[2, i].set_title("Overlay")
            axes[2, i].axis('off')

        plt.tight_layout()
        plt.show()

print('Disabling in-place ReLU modules...')
disable_inplace_relu(model)

sample_images = []
count = 0
for imgs, _ in test_loader:
    print(f"Batch of images shape: {imgs.shape}")  # For debug
    for img in imgs:
        print(f"Single image shape: {img.shape}")   # For debug
        sample_images.append(img)
        count += 1
        if count >= 5:
            break
    if count >= 5:
        break

background_samples = torch.stack([train_dataset[i][0] for i in range(10)])
print_shape_info("Raw background samples", background_samples)
background_samples = background_samples.to(DEVICE)
print_shape_info("Background samples on device", background_samples)

shap_explanation_batch(model, DEVICE, sample_images, background_samples, class_list, num_samples=5)
```
